```python
import numpy as np
import matplotlib.pyplot as plt


# Points class

class Point:
    """
    Represents a point in a 2D space

    attributes: x, y, name

    """

    # constructor function
    def __init__(self, x=0.0, y=0.0, name=""):
        self.x = x
        self.y = y
        self.name = name
        self.__private_variable = 42

    def __add__(self, other):
        assert isinstance(other, Point)

        new_point = Point(self.x + other.x, self.y + other.y)
        return new_point

    def print_point(self):
        print("Point ", self.name, "is", self.x, self.y)

    # Static function
    def calculate_distance_between_two_points(A, B):
        return np.sqrt((A.x - B.x)**2.0 + (A.y - B.y)**2.0)

    def move_point(self, dx, dy):
        self.x += dx
        self.y += dy

    # Reflect point in x-axis
    def reflect_x(self):
        self.y = -self.y

    # Reflect point in y-axis
    def reflect_y(self):
        self.x = -self.x

    def plot_point(self):
        plt.plot(self.x, self.y, 'o', label=self.name)

    def update_name(self, new_name):
        self.name = new_name


# Rectangle class

class Rectangle:
    """
    Represents a rectangle in a 2D space

    attributes: bottom_left, top_right

    """

    # constructor function
    def __init__(self, bottom_left, top_right):

        # Check that the corners are Point objects
        assert isinstance(bottom_left, Point)
        assert isinstance(top_right, Point)

        # Check that the rectangle has positive width and height
        assert bottom_left.x < top_right.x
        assert bottom_left.y < top_right.y

        self.bottom_left = bottom_left
        self.top_right = top_right

    # Reflect rectangle in x-axis
    def reflect_x(self):

        self.bottom_left.reflect_x()
        self.top_right.reflect_x()

        # Top and bottom have swapped
        new_bottom_left = Point(
            self.bottom_left.x,
            self.top_right.y
        )

        new_top_right = Point(
            self.top_right.x,
            self.bottom_left.y
        )

        self.bottom_left = new_bottom_left
        self.top_right = new_top_right

    # Reflect rectangle in y-axis
    def reflect_y(self):

        self.bottom_left.reflect_y()
        self.top_right.reflect_y()

        # Left and right have swapped
        new_bottom_left = Point(
            self.top_right.x,
            self.bottom_left.y
        )

        new_top_right = Point(
            self.bottom_left.x,
            self.top_right.y
        )

        self.bottom_left = new_bottom_left
        self.top_right = new_top_right

    # Grow rectangle with bottom left corner fixed
    def grow_bottom_left_fixed(self, dx, dy):

        assert dx >= 0
        assert dy >= 0

        self.top_right.x += dx
        self.top_right.y += dy

    # Grow rectangle with top right corner fixed
    def grow_top_right_fixed(self, dx, dy):

        assert dx >= 0
        assert dy >= 0

        self.bottom_left.x -= dx
        self.bottom_left.y -= dy

    # Plot rectangle
    def plot_rectangle(self):

        x = [
            self.bottom_left.x,
            self.top_right.x,
            self.top_right.x,
            self.bottom_left.x,
            self.bottom_left.x
        ]

        y = [
            self.bottom_left.y,
            self.bottom_left.y,
            self.top_right.y,
            self.top_right.y,
            self.bottom_left.y
        ]

        plt.plot(x, y)

    # Print rectangle
    def print_rectangle(self):

        print("Rectangle:")
        print("Bottom left corner:",
              self.bottom_left.x,
              self.bottom_left.y)

        print("Top right corner:",
              self.top_right.x,
              self.top_right.y)


# --------------------------------------------------
# Test the Rectangle class
# --------------------------------------------------

# Create two points
bottom_left = Point(1.0, 1.0, "Bottom Left")
top_right = Point(5.0, 4.0, "Top Right")

# Create rectangle
rectangle = Rectangle(bottom_left, top_right)

# Print rectangle
rectangle.print_rectangle()

# Plot rectangle
plt.plot()

rectangle.plot_rectangle()

plt.grid()
plt.xlabel("x position")
plt.ylabel("y position")
plt.title("Rectangle")

plt.show()

```

### ACTIVITY 3:

We now consider a simple system of coupled ODEs. The Lotka–Volterra equations describe predator–prey interactions between a prey population $x$ (for example, rabbits) and a predator population $y$ (for example, foxes). In the absence of predators, the prey population grows at a rate proportional to its size. In the absence of prey, the predator population declines at a rate proportional to its size. Predation occurs at a rate proportional to the product $xy$, reducing the prey population while supporting predator reproduction. The populations are therefore described by the following first-order, autonomous, nonlinear, coupled system of ordinary differential equations:

$\frac{dx}{dt} = ax - bxy$

$\frac{dy}{dt} = -cy + dxy$

Here, $a$ is the prey growth rate, $b$ measures predation, $c$ is the predator death rate, and $d$ measures how effectively consumed prey contributes to predator growth.

The following code solves the Logistic Equation

$\frac{dy}{dt} = y (1 - y)$

Using a simple explicit integration and the scipy function solve_ivp() which used Runge-Kutta 4th/5th order time integration. Copy it and update it to solve instead the Lotka-Volterra equations.

Note the following:

1. Why do I substract delta_t when calculating the vector of t_solution points?
2. the solve_ivp method uses the labels t and y for the independent and dependent variables. The y value is assumed to be a vector, allowing you to solve for multiple dependent variables at once, e.g. you can make it np.array([x,y])

```python
# Solve the 1d logistic equation from class
from scipy.integrate import solve_ivp

# Note that the function has to take t as the first argument and y as the second
def calculate_logistic_dydt(t, y):
    """Returns the gradient dx/dt for the logistic equation"""
    dydt = y*(1 - y)
    return dydt

max_time = 10.0
N_time_steps = 25
delta_t = max_time / N_time_steps
t_solution = np.linspace(0.0, max_time-delta_t, N_time_steps) # values of independent variable
y0 = np.array([0.5]) # an initial condition, y(0) = y0, note it needs to be an array

solution = solve_ivp(calculate_logistic_dydt, [0,max_time], y0,
                        method='RK45', t_eval=t_solution)

plt.grid()
plt.xlabel("t", fontsize=16)
plt.ylabel("y", fontsize=16)
plt.plot(solution.t, solution.y[0], 'o', label="solve_ivp")

# Now do it the "cheap" way
# increase the number of steps to see how the solution changes
y_solution = np.zeros_like(t_solution)
y_solution[0] = y0[0]
for itime, time in enumerate(t_solution) :
    if itime > 0 :
        dydt = calculate_logistic_dydt(time, y_solution[itime-1])
        y_solution[itime] = y_solution[itime-1] + dydt * delta_t

plt.plot(t_solution, y_solution, '-.',label="cheap method")

# Now plot the true solution
A = 1.0/y0 - 1.0
y_true = 1.0 / (1.0 + A * np.exp(-t_solution))
plt.plot(t_solution, y_true, '-', label="true solution")
plt.legend(loc='best');
```

```python
# Solve the Lotka-Volterra equations

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


# Define the coupled differential equations
def calculate_lotka_volterra_dydt(t, X):

    # X contains the two populations
    x = X[0]       # prey
    y = X[1]       # predator

    # Parameters
    a = 1.0        # prey growth rate
    b = 1.0        # predation rate
    c = 1.0        # predator death rate
    d = 1.0        # predator growth from prey

    # Differential equations
    dxdt = a*x - b*x*y
    dydt = -c*y + d*x*y

    # Return both derivatives
    return np.array([dxdt, dydt])


# Set up the time values
max_time = 20.0
N_time_steps = 500

delta_t = max_time / N_time_steps

t_solution = np.linspace(
    0.0,
    max_time - delta_t,
    N_time_steps
)


# Initial conditions
# X[0] = initial prey population
# X[1] = initial predator population

X0 = np.array([2.0, 1.0])


# Solve the equations

solution = solve_ivp(
    calculate_lotka_volterra_dydt,
    [0, max_time],
    X0,
    method='RK45',
    t_eval=t_solution
)


# Plot prey population

plt.plot(
    solution.t,
    solution.y[0],
    label="Prey"
)


# Plot predator population

plt.plot(
    solution.t,
    solution.y[1],
    label="Predator"
)


plt.grid()
plt.xlabel("t")
plt.ylabel("Population")
plt.legend(loc='best')

plt.show()
```

### ACTIVITY 6:

If you double the resolution in the simple integrator, what happens to the error in the solution compared to solve_ivp? Why? How could you improve this behaviour?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


# Solve the Lotka-Volterra equations

def calculate_lotka_volterra_dydt(t, X):
    """Returns dx/dt and dy/dt for the Lotka-Volterra equations"""

    x = X[0]
    y = X[1]

    a = 1.0
    b = 1.0
    c = 1.0
    d = 1.0

    dxdt = a*x - b*x*y
    dydt = -c*y + d*x*y

    return np.array([dxdt, dydt])


# Set up the time

max_time = 20.0

N_time_steps = 1000
delta_t = max_time / N_time_steps

t_solution = np.linspace(0.0, max_time-delta_t, N_time_steps)


# Initial conditions

X0 = np.array([2.0, 1.0])


# Solve using solve_ivp

solution = solve_ivp(
    calculate_lotka_volterra_dydt,
    [0, max_time],
    X0,
    method='RK45',
    t_eval=t_solution
)


# Simple Euler integrator

X_solution = np.zeros((2, N_time_steps))
X_solution[:, 0] = X0

for itime, time in enumerate(t_solution):

    if itime > 0:

        dXdt = calculate_lotka_volterra_dydt(
            time,
            X_solution[:, itime-1]
        )

        X_solution[:, itime] = (
            X_solution[:, itime-1] + dXdt * delta_t
        )


# ------------------------------------------------
# Plot the solutions
# ------------------------------------------------

plt.figure()

plt.plot(solution.t,
         solution.y[0],
         '-',
         label="solve_ivp prey")

plt.plot(t_solution,
         X_solution[0],
         '-.',
         label="Euler prey")

plt.plot(solution.t,
         solution.y[1],
         '-',
         label="solve_ivp predator")

plt.plot(t_solution,
         X_solution[1],
         '-.',
         label="Euler predator")

plt.grid()
plt.xlabel("t")
plt.ylabel("Population")
plt.legend(loc='best')
plt.show()


# ------------------------------------------------
# Compare the errors
# ------------------------------------------------

prey_error = solution.y[0] - X_solution[0]
predator_error = solution.y[1] - X_solution[1]

plt.figure()

plt.plot(solution.t,
         prey_error,
         '--',
         label=f"prey error, dt = {delta_t}")

plt.plot(solution.t,
         predator_error,
         '--',
         label=f"predator error, dt = {delta_t}")

plt.grid()
plt.xlabel("t")
plt.ylabel("Error")
plt.legend(loc='best')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


# Ecosystem class

class Ecosystem:
    """
    Represents a predator-prey ecosystem

    attributes: prey, predator, a, b, c, d
    """

    # Constructor
    def __init__(self, prey, predator, a, b, c, d):

        self.prey = prey
        self.predator = predator

        self.a = a
        self.b = b
        self.c = c
        self.d = d

    # Calculate dx/dt and dy/dt
    def calculate_dydt(self, t, X):

        x = X[0]
        y = X[1]

        dxdt = self.a*x - self.b*x*y
        dydt = -self.c*y + self.d*x*y

        return np.array([dxdt, dydt])

    # Solve the equations
    def solve(self, max_time, N_time_steps):

        delta_t = max_time / N_time_steps

        t_solution = np.linspace(
            0.0,
            max_time-delta_t,
            N_time_steps
        )

        X0 = np.array([self.prey, self.predator])

        solution = solve_ivp(
            self.calculate_dydt,
            [0, max_time],
            X0,
            method='RK45',
            t_eval=t_solution
        )

        return solution

    # Plot the populations
    def plot(self, solution):

        plt.plot(
            solution.t,
            solution.y[0],
            label="Prey"
        )

        plt.plot(
            solution.t,
            solution.y[1],
            label="Predator"
        )

        plt.grid()
        plt.xlabel("t")
        plt.ylabel("Population")
        plt.legend(loc='best')
        plt.show()


# Create an ecosystem

ecosystem = Ecosystem(
    2.0,    # initial prey population
    1.0,    # initial predator population
    1.0,    # prey growth rate (a)
    1.0,    # predation rate (b)
    1.0,    # predator death rate (c)
    1.0     # predator growth rate (d)
)


# Solve the equations

solution = ecosystem.solve(20.0, 500)


# Plot the solution

ecosystem.plot(solution)
```
